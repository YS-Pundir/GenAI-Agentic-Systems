```python
import openai
import json
import os 
from dotenv import load_dotenv
load_dotenv()
api_key=os.getenv("api_key")
```

```python
import os
from openai import OpenAI

# 1. Initialize the client pointing to Groq's endpoint
client = openai.OpenAI(
    api_key=api_key,
    base_url="https://api.groq.com/openai/v1"
)

response = client.responses.create(
    model="llama-3.3-70b-versatile",
    input="What models are trending on Huggingface?",
    tools=[
        {
            "type": "mcp",
            "server_label": "Huggingface",
            "server_url": "https://huggingface.co/mcp",
        }
    ]
)

print(response)

```

```python
import json

response_dict = response.model_dump() # Convert the response object to a dictionary
for item in response_dict['output']:
  if item['type'] == 'mcp_list_tools':
    print(f"Tools available from {item['server_label']}:")
    for tool in item['tools']:
      print(f"- {tool['name']}: {tool['description']}")
```

```python
import json

response_dict = response.model_dump() # Convert the response object to a dictionary
for item in response_dict['output']:
  if item['type'] == 'mcp_call':
    print("MCP Call Output:")
    print(item['output'])
```

```python
import json

response_dict = response.model_dump() # Convert the response object to a dictionary
for item in response_dict['output']:
  if item['type'] == 'message':
    print(item['content'][0]['text'])
```
