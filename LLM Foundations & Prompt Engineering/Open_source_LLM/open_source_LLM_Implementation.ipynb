```python
import os
from dotenv import load_dotenv

# This looks for the .env file in your current directory
load_dotenv()

# Access your API key using the exact name used in the .env file
# Replace 'YOUR_API_KEY_NAME' with the actual variable name (e.g., OPENAI_API_KEY)
api_key = os.getenv('api_key')

# Quick check (be careful not to print the whole key if sharing your screen!)
print(f"Key loaded: {api_key[:5]}...")
```
